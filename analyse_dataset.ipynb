```python
import pandas as pd

taxonomy = pd.read_csv("taxonomy.csv")

birds_taxonomy = taxonomy[taxonomy["class_name"] == "Aves"].copy()

birds_taxonomy[["primary_label", "common_name", "scientific_name", "class_name"]].head()
```

```python
print("Nombre total de classes :", len(taxonomy))
print("Nombre de classes d'oiseaux :", len(birds_taxonomy))
```

```python
from pathlib import Path

train_audio_dir = Path("train_audio")

rows = []

for _, row in birds_taxonomy.iterrows():
    label = row["primary_label"]
    folder = train_audio_dir / label

    if folder.exists():
        n_audio = len(list(folder.glob("*.ogg")))
        
        rows.append({
            "primary_label": label,
            "common_name": row["common_name"],
            "scientific_name": row["scientific_name"],
            "n_audio": n_audio
        })

birds_classes = pd.DataFrame(rows)

birds_classes = birds_classes.sort_values("n_audio", ascending=False)

birds_classes.head(20)
```

```python
selected_classes = birds_classes.head(6)

selected_classes
```

```python
selected_classes["primary_label"].to_csv(
    "classes.txt",
    index=False,
    header=False
)
```
