```python
import pandas as pd

split_df = pd.read_csv("splits/audio_split_65_15_20.csv")
classes_df = pd.read_csv("selected_classes.csv")

# split_df.head()
for i in classes_df["primary_label"] :
    df = split_df[split_df["label"] == i]
    count_tr = (df["split"] == "train").sum()
    count_val = (df["split"] == "val").sum()
    count_te = (df["split"] == "test").sum()
    print(f"Pour la classe {i}, il y a {count_tr} pour le train, {count_val} pour la validation, et {count_te} pour le test !")
    
```
